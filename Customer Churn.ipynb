**Correlation Analysis**

```python
df['Churn']
```

```python
df = df.drop("Customer_ID", axis=1)
```

```python
numeric_df = df.select_dtypes(include="number")

corr = numeric_df.corr()
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

The correlation analysis shows that Age, Tenure Months, Monthly Charges, and Support Calls have very weak linear correlations with Churn. Monthly Charges have the highest positive correlation with Churn (0.07), while Tenure Months has a very weak negative correlation (-0.01). Therefore, no strong linear relationship is observed between these numerical variables and customer churn.

 Data analysis reveals that customer churn is influenced more by categorical and behavioral factors than by simple demographics. Customers on monthly contracts are the most likely to leave, while those with one‑year or two‑year contracts show greater stability. DSL users churn slightly more than fiber users, suggesting service quality plays a role. Frequent support calls are strongly associated with churn, indicating dissatisfaction with customer service. Payment methods also affect churn, with certain options linked to higher exit rates. Interestingly, senior citizens are more loyal than younger customers, who churn at higher rates. Monthly charges show a weak positive correlation with churn, meaning higher bills slightly increase the risk of leaving, while longer tenure has a weak negative correlation, suggesting loyalty builds over time. Gender differences are minimal, showing churn is not strongly tied to male or female customers. Overall, the analysis highlights that contract type, internet service, support interactions, and payment methods are the most important drivers of churn, while age and tenure play smaller roles.
